```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import sys
from sklearn.preprocessing import StandardScaler
```

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:,1:2].values.astype(float)
y = dataset.iloc[:,2:3].values.astype(float)
```

```python
dataset
```

```python

sc_X = StandardScaler()

X = sc_X.fit_transform(X) #en este caso por que es solo una prueba, pero recuerda que si tuvieras un conjunto de entrenamiento y otro de prueba, deberías hacer fit_transform con el conjunto de entrenamiento y solo transform con el conjunto de prueba.
```

```python
from sklearn.svm import SVR
regressor = SVR(kernel='rbf')
regressor.fit(X,y)
valor_a_predecir = np.array([6.5]).reshape(-1, 1)
y_pred = regressor.predict(valor_a_predecir)

```

```python
plt.scatter(X, y, color = 'magenta')
plt.plot(X, regressor.predict(X), color = 'green')
plt.title('Truth or Bluff (Support Vector Regression Model)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```
